### Task for using the Child class 

```python
class Person:
    
    def __init__(self, name):
        self.name = name
    
    def show_name(self):
        print(f"Name is {self.name}")
        
class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name) #call Parent Constructor
        self.salary = salary 
        
    def show_salary(self):
        print(f"Salary is {self.salary}")
```

```python
e1 = Employee("Hamza", 50000)    
```

```python
e1.show_name()
```

```python
e1.show_salary()
```

```python
class Person:
    
    def __init__(self, name):
        self.name = name
        
    def show_name(self):
        print(f"Name is {self.name}")
        
class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary
        
    def show_Salary(self):
        print(f"Salary is {self.salary}")
```

```python
e1 = Employee("Hamza", 50000)
```

```python
e1.show_name()
```

```python
e1.show_Salary()
```

### Method overriding

```python
class Person:
    def show(self):
        print("Parent Call")
        
class Employee(Person):
    def show(self):
        print("Child Call")
```

```python
# When want to get parent function as well
class Employee(Person):
    def show(self):
        super().show()
        print("Child Call As Well")
```

```python
e1 = Employee()
```

```python
e1.show()
```

#### # Finding the second largest in array 

```python
arr = [12, 35, 1, 10, 34, 1]
```

```python
max_value = max(arr)
max_value
```

```python
sec_lar = [x for x in arr if x != max_value]
```

```python
sec_lar
```

```python
arr = [12, 35, 1, 10, 34, 1]

first = max(arr)
arr2 = [x for x in arr if x != first]

second = max(arr2)
print(second)
```

### Finding the repeated item in an array 

```python
arr = [1, 5, 3, 4, 3, 5, 6] 
```

```python
# finding repeated items using list comprehension
repeated_items = list(set([x for x in arr if arr.count(x)>1]))
```

```python
repeated_items
```

```python
chk_list = [1, 2, 3, 4]
```

```python
list(set([x for x in chk_list if chk_list.count(x)>1]))
```

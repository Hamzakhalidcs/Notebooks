```python
#  Step 1: Base Class Person

class Person:
    def __init__(self, name):
        self.name = name
    
    def show(self):
        print(f"Name is {self.name}")
```

```python
# step 2: Student Class (Inheritacne+Logic)

class Student(Person):
    
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks
        
    def result(self):
        return "PASS" if self.marks>50 else "FAIL"
    
    def grade(self):
        
        if self.marks>=80:
            return "A"
        
        elif self.marks>=60:
            return "B"
        
        elif self.marks>=50:
            return "C"
        
        else:
            return "FAIL"
        
    def show(self):
        super().show()
        print(f"Marks : {self.marks}")
        
```

```python
# Step 3: Create Multiple Students

students = [
    Student("Hamza", 75),
    Student("Ali", 45),
    Student("Sara", 85)
]
```

```python
# Step 4: Loop through Students
for s in students:
    print("*"*20)
    s.show()
    print("Result", s.result())
    print("Grade", s.grade())
```
